## Mathematical example

This separable fixture checks optimization and class decoding, not speech performance.


```python
x=np.array([[-2.,-1.],[-1.,-2.],[2.,1.],[1.,2.]])
y=np.array([-3,-3,50,50])
model=LinearSVM(n_iters=500).fit(x,y)
assert np.array_equal(model.predict(x),y)
```
